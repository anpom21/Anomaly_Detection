```python
import numpy as np

import matplotlib.pyplot as plt

# Generate 1000 random 2D points from a uniform distribution
np.random.seed(42)  # For reproducibility
n = 5000
points = np.random.uniform(low=0, high=100, size=(n, 2))

plt.rcParams['text.usetex'] = True
plt.rcParams['font.family'] = 'serif'  # LaTeX default font is Computer Modern (serif)
# Randomly select 2% (20 points) and 10% (100 points) subsets
indices = np.arange(n)
subset_2_idx = np.random.choice(indices, size=int(0.02 * n), replace=False)
subset_10_idx = np.random.choice(indices, size=int(0.10 * n), replace=False)

# Plot with 2% subset in red
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.scatter(points[:, 0], points[:, 1], color='blue', alpha=0.5, label='All points')
plt.scatter(points[subset_2_idx, 0], points[subset_2_idx, 1], color='red', label='2\% subset')
plt.title('2\% Coverage')
plt.legend()
plt.xlabel('')
plt.ylabel('')
plt.xticks([])
plt.yticks([])


# Plot with 10% subset in red
plt.subplot(1, 2, 2)
plt.scatter(points[:, 0], points[:, 1], color='blue', alpha=0.5, label='All points')
plt.scatter(points[subset_10_idx, 0], points[subset_10_idx, 1], color='red', label='10\% subset')
plt.title('10\% Coverage')
plt.legend()
plt.xlabel('')
plt.ylabel('')
plt.xticks([])
plt.yticks([])


plt.tight_layout()
plt.show()
```
